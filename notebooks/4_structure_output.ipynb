### Structure Output

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm =  ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
from pydantic import BaseModel, Field
from typing import List
```

```python
class ResponseStrcuture(BaseModel):
    name: str = Field(..., description="Name of the person")
    email: str = Field(..., description="Email of the person")
    age: int = Field(..., description="Age of the person")
    
structured_output = llm.with_structured_output(ResponseStrcuture)
```

```python
res = structured_output.invoke("Please provide your name, email, and age.")
```

```python
res.email
```

```python
class movie(BaseModel):
    title: str = Field(..., description="Title of the movie")
    year: int = Field(..., description="Year of release")
    genre: str = Field(..., description="Genre of the movie")

class AllMovies(BaseModel):
    movies: List[movie]

```

```python
movie_llm = llm.with_structured_output(AllMovies)
```

```python
movie_llm.invoke("Give me latest 3 movie name  ")
```
