```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_number(a: int, b: int) -> int:
    """
    It will Return The sum of two numbers.
    Args:
        a (int): The first number.
        b (int): The second number. 
    """
    return a + b

@tool
def product_number(a: int, b: int) -> int:
    """
    It will Return The product of two numbers.
    Args:
        a (int): The first number.
        b (int): The second number. 
    """
    return a * b
```

```python
add_number.invoke({"a": 5, "b": 10})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b", temperature=0.3, streaming=True)
```

```python
agent = create_agent(
    model=llm,
    tools=[add_number, product_number],
    system_prompt="You are a helpful assistant that can perform addition and multiplication of two numbers. You have access to the following tools: add_number and product_number. Use these tools to answer questions about addition and multiplication."
    )
```

```python
response = agent.invoke({"messages":[{"role":"user","content":"What is the sum of 5 and 10 * 20?"}]})
```

```python
print(response["messages"][-1].text)
```

```python
for res in response['messages']:
    print(res)
    print("\n")
```
