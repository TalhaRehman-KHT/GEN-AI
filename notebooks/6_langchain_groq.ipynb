```python
from dotenv import load_dotenv
load_dotenv()
```

```python
### using groq with langchain
from langchain_groq import ChatGroq

### using google gemini with langchain
from langchain_google_genai import ChatGoogleGenerativeAI


```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    )

gemini_llm =  ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
## responce time with groq 
res= llm.invoke("Write a poem about a lonely robot in the style of Shakespeare.")
```

```python
### response time with google gemini
result_with_gemini = gemini_llm.invoke("Write a poem about a lonely robot in the style of Shakespeare.")
```
