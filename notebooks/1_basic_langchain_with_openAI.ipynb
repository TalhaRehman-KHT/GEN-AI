```python
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq
from langchain_anthropic import ChatAnthropic

openai_llm = ChatOpenAI(
	model="gpt-5-mini-2025-08-07",
	api_key=os.getenv("OPEN_API_KEY") or os.getenv("OPEN_API_KEY"),
)
gemini_llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
groq_llm   = ChatGroq(model="llama-3.3-70b-versatile")
claude_llm = ChatAnthropic(model="claude-haiku-4-5-20251001")
```

```python
models = {
    "OpenAI": openai_llm,
    "Gemini": gemini_llm,
    "Groq": groq_llm,
    "Claude": claude_llm,
}

for name, model in models.items():
    try:
        r = model.invoke("What is the capital of France?")
        print(f"{name}: {r.text}")
    except Exception as e:
        print(f"{name} failed: {type(e).__name__}: {e}")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
prompts = [
    ("system" , "You are a python Developer."),
    ("user" , "Write a python code to sort array ."),
]
res = llm.invoke(prompts)
print(res.content)
```
