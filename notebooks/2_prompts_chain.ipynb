### Dynamic Prompts / Prompts Templates

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser


out = StrOutputParser()
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite")
```

```python
prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a translator. Translate the user's message into {language}. Reply with the translation only."),
    ("human", "{query}"),
])
```

```python
final_prompt = prompts.format_prompt(language="Urdu", query="Hello, how are you?")
print(final_prompt)
```

```python
res =llm.invoke(final_prompt)
```

```python
print(res.content)
```

```python
def tranform_uppercase(result: str):
    return result.upper()
```

```python
chain = prompts | llm | out | tranform_uppercase
```

```python
res = chain.invoke({"language": "Hinglsih", "query": "Hello, how are you?"})
print(res)
```
